```python
import numpy as np

# Define the activation function and its derivative
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return sigmoid(x) * (1 - sigmoid(x))

# Mean Squared Error Loss Function
def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

# Initialize parameters
input_size = 1  # Number of input features
hidden_size = 4  # Number of neurons in the hidden layer
output_size = 1  # Number of output neurons
learning_rate = 0.01
epochs = 1000

# Initialize weights and biases
np.random.seed(42)
weights_input_hidden = np.random.randn(input_size, hidden_size)
weights_hidden_output = np.random.randn(hidden_size, output_size)
bias_hidden = np.zeros(hidden_size)
bias_output = np.zeros(output_size)

# Training data
X_train = np.array([[0], [1], [2], [3], [4]])
y_train = np.array([[0], [1], [4], [9], [16]])

# Training loop
for epoch in range(epochs):
    # Forward propagation
    hidden_layer_input = np.dot(X_train, weights_input_hidden) + bias_hidden
    hidden_layer_output = sigmoid(hidden_layer_input)
    output_layer_input = np.dot(hidden_layer_output, weights_hidden_output) + bias_output
    y_pred = output_layer_input
    
    # Compute loss
    loss = mean_squared_error(y_train, y_pred)
    
    # Backward propagation
    output_error = y_pred - y_train
    output_gradient = output_error
    
    hidden_error = np.dot(output_gradient, weights_hidden_output.T) * sigmoid_derivative(hidden_layer_input)
    hidden_gradient = np.dot(X_train.T, hidden_error)
    
    # Update weights and biases
    weights_hidden_output -= learning_rate * np.dot(hidden_layer_output.T, output_gradient)
    bias_output -= learning_rate * np.sum(output_gradient, axis=0)
    weights_input_hidden -= learning_rate * hidden_gradient
    bias_hidden -= learning_rate * np.sum(hidden_error, axis=0)
    
    if epoch % 100 == 0:
        print(f'Epoch {epoch}, Loss: {loss}')

# Final predictions
print("\nFinal Predictions:")
print(y_pred)
```
